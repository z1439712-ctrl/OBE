```python

# ================================================================
# BANK QUEUE & CUSTOMER SERVICE SIMULATION
# Simulation and Modeling Lab - Complete Project
# Python + SimPy + NumPy + Pandas + Matplotlib
# ================================================================

# CELL 1: INSTALL LIBRARIES
!pip -q install simpy numpy pandas matplotlib

# CELL 2: IMPORT LIBRARIES
import simpy
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
import zipfile
import textwrap

from google.colab import files
from IPython.display import display, Markdown

print("All Libraries Imported Successfully!")

# ================================================================
# CELL 3: PROJECT CONFIGURATION
# ================================================================

SIM_TIME = 480
AVG_ARRIVAL_TIME = 2
AVG_SERVICE_TIME = 6

MIN_TELLERS = 1
MAX_TELLERS = 10
REPLICATIONS = 5

TELLER_COST_PER_HOUR = 300
WAITING_COST_PER_CUSTOMER_HOUR = 50

OUTPUT_DIR = "bank_simulation_outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

print("\nBANK SIMULATION CONFIGURATION")
print("=" * 50)
print("Bank Operating Time:", SIM_TIME, "Minutes")
print("Average Arrival Interval:", AVG_ARRIVAL_TIME, "Minutes")
print("Average Service Time:", AVG_SERVICE_TIME, "Minutes")
print("Teller Range:", MIN_TELLERS, "to", MAX_TELLERS)
print("Simulation Replications:", REPLICATIONS)

# ================================================================
# CELL 4: CUSTOMER SERVICE PROCESS
# ================================================================

def customer(env, customer_id, bank, customer_results):

    arrival_time = env.now

    with bank.request() as request:

        yield request

        service_start = env.now
        waiting_time = service_start - arrival_time

        service_time = random.expovariate(
            1 / AVG_SERVICE_TIME
        )

        yield env.timeout(service_time)

        departure_time = env.now

        customer_results.append({

            "Customer ID": customer_id,
            "Arrival Time": arrival_time,
            "Service Start": service_start,
            "Waiting Time": waiting_time,
            "Service Time": service_time,
            "Departure Time": departure_time

        })


# ================================================================
# CELL 5: CUSTOMER ARRIVAL PROCESS
# ================================================================

def customer_arrivals(env, bank, customer_results):

    customer_id = 0

    while True:

        arrival_interval = random.expovariate(
            1 / AVG_ARRIVAL_TIME
        )

        yield env.timeout(arrival_interval)

        # No new arrivals after bank closing
        if env.now >= SIM_TIME:
            break

        customer_id += 1

        env.process(
            customer(
                env,
                customer_id,
                bank,
                customer_results
            )
        )


# ================================================================
# CELL 6: QUEUE MONITORING
# ================================================================

def monitor_queue(env, bank, queue_data):

    while env.now < SIM_TIME:

        queue_data.append({

            "Time": env.now,
            "Queue Length": len(bank.queue),
            "Busy Tellers": bank.count

        })

        yield env.timeout(1)


# ================================================================
# CELL 7: MAIN SIMULATION FUNCTION
# ================================================================

def run_simulation(number_of_tellers, seed):

    random.seed(seed)
    np.random.seed(seed)

    env = simpy.Environment()

    bank = simpy.Resource(
        env,
        capacity=number_of_tellers
    )

    customer_results = []
    queue_data = []

    env.process(
        customer_arrivals(
            env,
            bank,
            customer_results
        )
    )

    env.process(
        monitor_queue(
            env,
            bank,
            queue_data
        )
    )

    # Run during banking hours
    env.run(until=SIM_TIME)

    # Complete all customers who arrived before closing
    env.run()

    customers_df = pd.DataFrame(customer_results)
    queue_df = pd.DataFrame(queue_data)

    total_customers = len(customers_df)

    if total_customers > 0:

        average_wait = customers_df[
            "Waiting Time"
        ].mean()

        total_wait_minutes = customers_df[
            "Waiting Time"
        ].sum()

        average_service = customers_df[
            "Service Time"
        ].mean()

    else:

        average_wait = 0
        total_wait_minutes = 0
        average_service = 0

    average_queue = (
        queue_df["Queue Length"].mean()
        if not queue_df.empty else 0
    )

    # Calculate teller busy time during opening hours
    busy_minutes = 0

    for row in customer_results:

        start = row["Service Start"]
        end = row["Departure Time"]

        overlap_start = max(0, start)
        overlap_end = min(SIM_TIME, end)

        busy_minutes += max(
            0,
            overlap_end - overlap_start
        )

    utilization = (
        busy_minutes /
        (number_of_tellers * SIM_TIME)
    ) * 100

    # Operational cost
    teller_cost = (
        number_of_tellers *
        (SIM_TIME / 60) *
        TELLER_COST_PER_HOUR
    )

    waiting_cost = (
        total_wait_minutes / 60
    ) * WAITING_COST_PER_CUSTOMER_HOUR

    total_cost = teller_cost + waiting_cost

    return {

        "Customers": total_customers,
        "Average Waiting Time": average_wait,
        "Average Service Time": average_service,
        "Average Queue Length": average_queue,
        "Teller Utilization (%)": utilization,
        "Teller Cost (BDT)": teller_cost,
        "Waiting Cost (BDT)": waiting_cost,
        "Total Cost (BDT)": total_cost

    }


# ================================================================
# CELL 8: RUN ALL EXPERIMENTS
# ================================================================

all_results = []

print("\nBANK QUEUE SIMULATION STARTED")
print("=" * 90)

for tellers in range(
    MIN_TELLERS,
    MAX_TELLERS + 1
):

    replication_results = []

    for rep in range(REPLICATIONS):

        result = run_simulation(
            tellers,
            seed=1000 + rep
        )

        replication_results.append(result)

    result_df = pd.DataFrame(
        replication_results
    )

    final_result = result_df.mean(
        numeric_only=True
    ).to_dict()

    final_result["Tellers"] = tellers

    all_results.append(final_result)

    print(
        f"Tellers: {tellers:2d} | "
        f"Customers: {final_result['Customers']:.0f} | "
        f"Waiting: {final_result['Average Waiting Time']:.2f} min | "
        f"Queue: {final_result['Average Queue Length']:.2f} | "
        f"Utilization: {final_result['Teller Utilization (%)']:.2f}% | "
        f"Cost: {final_result['Total Cost (BDT)']:.2f} BDT"
    )

results_df = pd.DataFrame(
    all_results
).sort_values(
    "Tellers"
).reset_index(drop=True)

print("\nSIMULATION COMPLETED SUCCESSFULLY!")


# ================================================================
# CELL 9: DISPLAY FINAL COMPARISON TABLE
# ================================================================

print("\nFINAL SIMULATION COMPARISON TABLE")
display(results_df.round(2))


# ================================================================
# CELL 10: OPTIMAL TELLER ANALYSIS
# ================================================================

best_result = results_df.loc[
    results_df["Total Cost (BDT)"].idxmin()
]

print("\n" + "=" * 60)
print("OPTIMAL BANK COUNTER ANALYSIS")
print("=" * 60)

print(
    "Optimal Number of Tellers:",
    int(best_result["Tellers"])
)

print(
    "Average Waiting Time:",
    round(best_result["Average Waiting Time"], 2),
    "Minutes"
)

print(
    "Average Queue Length:",
    round(best_result["Average Queue Length"], 2)
)

print(
    "Teller Utilization:",
    round(best_result["Teller Utilization (%)"], 2),
    "%"
)

print(
    "Minimum Total Cost:",
    round(best_result["Total Cost (BDT)"], 2),
    "BDT"
)


# ================================================================
# CELL 11: GRAPH 1 - AVERAGE WAITING TIME
# ================================================================

plt.figure(figsize=(10, 6))

plt.plot(
    results_df["Tellers"],
    results_df["Average Waiting Time"],
    marker="o",
    linewidth=2
)

plt.xlabel("Number of Tellers")
plt.ylabel("Average Waiting Time (Minutes)")
plt.title("Average Waiting Time vs Number of Tellers")

plt.xticks(range(1, 11))
plt.grid(True)
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR + "/waiting_time.png",
    dpi=300
)

plt.show()


# ================================================================
# CELL 12: GRAPH 2 - TOTAL OPERATIONAL COST
# ================================================================

plt.figure(figsize=(10, 6))

plt.plot(
    results_df["Tellers"],
    results_df["Total Cost (BDT)"],
    marker="s",
    linewidth=2
)

plt.xlabel("Number of Tellers")
plt.ylabel("Total Cost (BDT)")
plt.title("Total Operational Cost vs Number of Tellers")

plt.xticks(range(1, 11))
plt.grid(True)
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR + "/cost_comparison.png",
    dpi=300
)

plt.show()


# ================================================================
# CELL 13: GRAPH 3 - AVERAGE QUEUE LENGTH
# ================================================================

plt.figure(figsize=(10, 6))

plt.plot(
    results_df["Tellers"],
    results_df["Average Queue Length"],
    marker="^",
    linewidth=2
)

plt.xlabel("Number of Tellers")
plt.ylabel("Average Queue Length")
plt.title("Average Queue Length vs Number of Tellers")

plt.xticks(range(1, 11))
plt.grid(True)
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR + "/queue_comparison.png",
    dpi=300
)

plt.show()


# ================================================================
# CELL 14: GRAPH 4 - TELLER UTILIZATION
# ================================================================

plt.figure(figsize=(10, 6))

plt.plot(
    results_df["Tellers"],
    results_df["Teller Utilization (%)"],
    marker="D",
    linewidth=2
)

plt.xlabel("Number of Tellers")
plt.ylabel("Teller Utilization (%)")
plt.title("Teller Utilization vs Number of Tellers")

plt.xticks(range(1, 11))
plt.grid(True)
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR + "/teller_utilization.png",
    dpi=300
)

plt.show()


# ================================================================
# CELL 15: CUSTOMER AND QUEUE PERFORMANCE ANALYSIS
# ================================================================

print("\nPERFORMANCE ANALYSIS")
print("=" * 60)

print(
    "Total Average Customers:",
    round(results_df["Customers"].mean(), 2)
)

print(
    "Lowest Waiting Time:",
    round(results_df["Average Waiting Time"].min(), 2),
    "Minutes"
)

print(
    "Highest Waiting Time:",
    round(results_df["Average Waiting Time"].max(), 2),
    "Minutes"
)

print(
    "Lowest Queue Length:",
    round(results_df["Average Queue Length"].min(), 2)
)

print(
    "Highest Queue Length:",
    round(results_df["Average Queue Length"].max(), 2)
)


# ================================================================
# CELL 16: MATHEMATICAL MODEL
# ================================================================

arrival_rate = 1 / AVG_ARRIVAL_TIME
service_rate = 1 / AVG_SERVICE_TIME

print("\nMATHEMATICAL MODEL")
print("=" * 60)

print("Arrival Rate (Lambda):", arrival_rate)
print("Service Rate (Mu):", service_rate)

print("\nTraffic Intensity:")

for c in range(MIN_TELLERS, MAX_TELLERS + 1):

    rho = arrival_rate / (c * service_rate)

    print(
        f"{c} Teller(s): {rho:.3f} "
        f"({rho * 100:.2f}%)"
    )

print("\nFormula:")
print("Lambda = 1 / Average Arrival Time")
print("Mu = 1 / Average Service Time")
print("Rho = Lambda / (Number of Tellers * Mu)")
print("Total Cost = Teller Cost + Waiting Cost")


# ================================================================
# CELL 17: SYSTEM ARCHITECTURE DIAGRAM
# ================================================================

from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

def draw_architecture():

    fig, ax = plt.subplots(figsize=(13, 7))

    ax.set_xlim(0, 12)
    ax.set_ylim(0, 8)
    ax.axis("off")

    boxes = [
        (0.5, 5.5, 2.2, 1.1, "INPUT LAYER\nSimulation Parameters"),
        (3.4, 5.5, 2.2, 1.1, "ARRIVAL MODULE\nCustomer Generation"),
        (6.3, 5.5, 2.2, 1.1, "QUEUE MODULE\nSimPy Resource"),
        (9.2, 5.5, 2.2, 1.1, "SERVICE MODULE\nTeller Processing"),
        (3.4, 2.5, 2.2, 1.1, "DATA COLLECTION\nCustomer Records"),
        (6.3, 2.5, 2.2, 1.1, "ANALYSIS MODULE\nPerformance & Cost"),
        (9.2, 2.5, 2.2, 1.1, "OUTPUT LAYER\nGraphs & CSV")
    ]

    for x, y, w, h, label in boxes:

        box = FancyBboxPatch(
            (x, y), w, h,
            boxstyle="round,pad=0.12",
            linewidth=1.5,
            edgecolor="black",
            facecolor="white"
        )

        ax.add_patch(box)

        ax.text(
            x + w/2,
            y + h/2,
            label,
            ha="center",
            va="center",
            fontsize=9,
            weight="bold"
        )

    arrows = [
        ((2.7, 6.05), (3.4, 6.05)),
        ((5.6, 6.05), (6.3, 6.05)),
        ((8.5, 6.05), (9.2, 6.05)),
        ((10.3, 5.5), (4.5, 3.6)),
        ((5.6, 3.05), (6.3, 3.05)),
        ((8.5, 3.05), (9.2, 3.05))
    ]

    for start, end in arrows:

        ax.add_patch(
            FancyArrowPatch(
                start,
                end,
                arrowstyle="->",
                mutation_scale=15,
                linewidth=1.5
            )
        )

    ax.set_title(
        "BANK QUEUE SIMULATION - SYSTEM ARCHITECTURE",
        fontsize=14,
        weight="bold"
    )

    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR + "/system_architecture.png",
        dpi=300
    )

    plt.show()


draw_architecture()


# ================================================================
# CELL 18: SIMULATION FLOWCHART
# ================================================================

def draw_flowchart():

    fig, ax = plt.subplots(figsize=(10, 13))

    ax.set_xlim(0, 10)
    ax.set_ylim(0, 15)
    ax.axis("off")

    steps = [
        "START",
        "Set Simulation Parameters",
        "Select Number of Tellers",
        "Initialize SimPy Environment",
        "Generate Customer Arrivals",
        "Customer Requests Teller",
        "Wait if Teller is Busy",
        "Receive Customer Service",
        "Record Simulation Data",
        "Bank Closing Time",
        "Calculate Performance Metrics",
        "Calculate Total Operational Cost",
        "Compare Teller Configurations",
        "Select Minimum Cost Configuration",
        "END"
    ]

    y_positions = np.linspace(14, 1, len(steps))

    for i, (step, y) in enumerate(
        zip(steps, y_positions)
    ):

        box = FancyBboxPatch(
            (2.2, y - 0.25),
            5.6,
            0.55,
            boxstyle="round,pad=0.08",
            edgecolor="black",
            facecolor="white"
        )

        ax.add_patch(box)

        ax.text(
            5,
            y,
            step,
            ha="center",
            va="center",
            fontsize=9
        )

        if i < len(steps) - 1:

            ax.annotate(
                "",
                xy=(5, y_positions[i+1] + 0.3),
                xytext=(5, y - 0.3),
                arrowprops=dict(
                    arrowstyle="->",
                    linewidth=1.2
                )
            )

    ax.set_title(
        "BANK QUEUE SIMULATION - FLOWCHART",
        fontsize=14,
        weight="bold"
    )

    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR + "/simulation_flowchart.png",
        dpi=300
    )

    plt.show()


draw_flowchart()


# ================================================================
# CELL 19: DFD LEVEL 0
# ================================================================

def draw_dfd_level0():

    fig, ax = plt.subplots(figsize=(11, 5))

    ax.set_xlim(0, 12)
    ax.set_ylim(0, 6)
    ax.axis("off")

    ax.text(
        1.5, 3,
        "PROJECT USER",
        ha="center",
        va="center",
        bbox=dict(boxstyle="round", fc="white", ec="black"),
        fontsize=11
    )

    circle = plt.Circle(
        (6, 3), 1.3,
        fill=False,
        linewidth=2
    )

    ax.add_patch(circle)

    ax.text(
        6, 3,
        "BANK QUEUE\nSIMULATION\nSYSTEM",
        ha="center",
        va="center",
        fontsize=10,
        weight="bold"
    )

    ax.text(
        10.2, 3,
        "RESULTS",
        ha="center",
        va="center",
        bbox=dict(boxstyle="round", fc="white", ec="black"),
        fontsize=11
    )

    ax.annotate(
        "Simulation Parameters",
        xy=(4.7, 3.5),
        xytext=(2.2, 4.5),
        arrowprops=dict(arrowstyle="->"),
        ha="center"
    )

    ax.annotate(
        "Performance Reports",
        xy=(8.5, 3.5),
        xytext=(10, 4.5),
        arrowprops=dict(arrowstyle="->"),
        ha="center"
    )

    ax.annotate(
        "Feedback",
        xy=(1.8, 2.5),
        xytext=(4.5, 1.5),
        arrowprops=dict(arrowstyle="->"),
        ha="center"
    )

    ax.set_title(
        "DATA FLOW DIAGRAM - LEVEL 0",
        fontsize=14,
        weight="bold"
    )

    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR + "/dfd_level_0.png",
        dpi=300
    )

    plt.show()


draw_dfd_level0()


# ================================================================
# CELL 20: DFD LEVEL 1
# ================================================================

def draw_dfd_level1():

    fig, ax = plt.subplots(figsize=(13, 7))

    ax.set_xlim(0, 14)
    ax.set_ylim(0, 8)
    ax.axis("off")

    nodes = [
        (0.5, 5, "USER"),
        (3, 5, "1. CONFIGURATION"),
        (6, 5, "2. CUSTOMER ARRIVAL"),
        (9, 5, "3. QUEUE & SERVICE"),
        (12, 5, "4. PERFORMANCE\nANALYSIS"),
        (9, 1.8, "5. COST ANALYSIS"),
        (12, 1.8, "6. FINAL OUTPUT")
    ]

    for x, y, label in nodes:

        box = FancyBboxPatch(
            (x, y),
            1.6,
            0.9,
            boxstyle="round,pad=0.1",
            linewidth=1.5,
            edgecolor="black",
            facecolor="white"
        )

        ax.add_patch(box)

        ax.text(
            x + 0.8,
            y + 0.45,
            label,
            ha="center",
            va="center",
            fontsize=8,
            weight="bold"
        )

    connections = [
        ((2.1, 5.45), (3, 5.45)),
        ((4.6, 5.45), (6, 5.45)),
        ((7.6, 5.45), (9, 5.45)),
        ((10.6, 5.45), (12, 5.45)),
        ((12.8, 5), (10, 2.7)),
        ((10.6, 2.25), (12, 2.25))
    ]

    for start, end in connections:

        ax.annotate(
            "",
            xy=end,
            xytext=start,
            arrowprops=dict(
                arrowstyle="->",
                linewidth=1.5
            )
        )

    ax.set_title(
        "DATA FLOW DIAGRAM - LEVEL 1",
        fontsize=14,
        weight="bold"
    )

    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR + "/dfd_level_1.png",
        dpi=300
    )

    plt.show()


draw_dfd_level1()


# ================================================================
# CELL 21: SAVE COMPLETE SIMULATION RESULTS
# ================================================================

csv_path = OUTPUT_DIR + "/bank_simulation_results.csv"

results_df.to_csv(
    csv_path,
    index=False
)

print("CSV Results Saved Successfully!")


# ================================================================
# CELL 22: PROJECT REPORT CONTENT
# ================================================================

report_content = f"""
BANK QUEUE & CUSTOMER SERVICE SIMULATION

1. INTRODUCTION

This project simulates a bank queue system using Python
and SimPy. The main purpose is to analyze customer waiting
time and teller performance.

2. PROBLEM STATEMENT

Banks often experience long customer queues. An insufficient
number of tellers can increase waiting time, while too many
tellers can increase operational expenses.

3. OBJECTIVES

- Simulate customer arrivals.
- Model customer waiting queues.
- Analyze teller service performance.
- Calculate average waiting time.
- Calculate teller utilization.
- Estimate operational costs.
- Identify the optimal number of tellers.

4. METHODOLOGY

The project uses discrete-event simulation.
Customer arrival and service times follow exponential
distributions.

5. SIMULATION CONFIGURATION

Operating Time: {SIM_TIME} minutes
Average Arrival Interval: {AVG_ARRIVAL_TIME} minutes
Average Service Time: {AVG_SERVICE_TIME} minutes
Teller Range: {MIN_TELLERS}-{MAX_TELLERS}
Replications: {REPLICATIONS}

6. OPTIMAL RESULT

Optimal Tellers: {int(best_result['Tellers'])}
Average Waiting Time: {best_result['Average Waiting Time']:.2f} minutes
Average Queue Length: {best_result['Average Queue Length']:.2f}
Teller Utilization: {best_result['Teller Utilization (%)']:.2f}%
Minimum Estimated Cost: {best_result['Total Cost (BDT)']:.2f} BDT

7. MATHEMATICAL MODEL

Lambda = 1 / Average Arrival Time
Mu = 1 / Average Service Time
Rho = Lambda / (Number of Tellers * Mu)

Total Cost = Teller Cost + Waiting Cost

8. CONCLUSION

The simulation compares different teller configurations
and evaluates their performance using waiting time,
queue length, utilization and estimated total cost.

The minimum-cost configuration is selected as the
recommended teller configuration under the assumptions
used in this experiment.
"""

report_path = OUTPUT_DIR + "/project_report_content.txt"

with open(report_path, "w", encoding="utf-8") as f:
    f.write(report_content)

print("\nPROJECT REPORT CONTENT CREATED!")
print(report_content)


# ================================================================
# CELL 23: ASSUMPTIONS AND LIMITATIONS
# ================================================================

assumptions = """
PROJECT ASSUMPTIONS AND LIMITATIONS

1. Customer arrivals follow an exponential distribution.
2. Service times follow an exponential distribution.
3. All tellers have the same service rate.
4. Customers follow First Come First Served (FCFS).
5. Customers do not abandon the queue.
6. Teller breaks and downtime are not considered.
7. Cost values are assumed for academic purposes.
8. Customers arriving before closing can finish service
   after the bank closes.
9. Only five replications are used in this experiment.
"""

print(assumptions)

with open(
    OUTPUT_DIR + "/assumptions.txt",
    "w",
    encoding="utf-8"
) as f:
    f.write(assumptions)


# ================================================================
# CELL 24: FINAL PROJECT SUMMARY
# ================================================================

print("\n" + "=" * 70)
print("BANK QUEUE SIMULATION - FINAL PROJECT SUMMARY")
print("=" * 70)

print("Project Status: COMPLETED")
print("Total Teller Configurations:", MAX_TELLERS)
print("Replications per Configuration:", REPLICATIONS)
print("Total Experiments:", MAX_TELLERS * REPLICATIONS)

print(
    "Optimal Teller Configuration:",
    int(best_result["Tellers"])
)

print(
    "Minimum Estimated Total Cost:",
    round(best_result["Total Cost (BDT)"], 2),
    "BDT"
)

print("\nGenerated Project Files:")

for filename in sorted(os.listdir(OUTPUT_DIR)):
    print("-", filename)


# ================================================================
# CELL 25: CREATE COMPLETE PROJECT ZIP
# ================================================================

zip_path = "Bank_Queue_Complete_Project.zip"

with zipfile.ZipFile(
    zip_path,
    "w",
    zipfile.ZIP_DEFLATED
) as zipf:

    for root, dirs, filenames in os.walk(OUTPUT_DIR):

        for filename in filenames:

            full_path = os.path.join(
                root,
                filename
            )

            zipf.write(
                full_path,
                arcname=os.path.relpath(
                    full_path,
                    OUTPUT_DIR
                )
            )

print("\nCOMPLETE PROJECT ZIP CREATED!")
print("File:", zip_path)


# ================================================================
# CELL 26: DOWNLOAD COMPLETE PROJECT
# ================================================================

files.download(zip_path)

print("\nALL PROJECT TASKS COMPLETED SUCCESSFULLY!")
print("Thank you!")
```
